# The Wolff cluster algorithm

Single-spin samplers (heat bath, Metropolis) slow down dramatically near the critical temperature $kT_c$.
Large aligned regions form, and flipping one spin at a time takes a very long time to change them.
The Wolff algorithm instead flips a whole cluster of aligned spins in a single step:

1. **Seed**: pick a random spin.
2. **Grow**: for each bond between a cluster spin and a parallel neighbor outside the cluster, add that neighbor with probability $p = 1 - e^{-2J/kT}$.
   Each bond gets its own chance, so a neighbor touching several cluster spins is more likely to join.
   Repeat for the newly added spins until no more join.
3. **Flip**: flip every spin in the cluster.

This choice of $p$ makes the algorithm sample the Boltzmann distribution exactly, with every cluster flip accepted.

Below, each click of **Next Step** shows one stage: the seed (dark orange), then its cluster (light orange), then the grid after the flip.
The demo runs at $kT_c$.
The bottom plot tracks the correlation between the current grid and the initial grid: the faster it decays, the faster the algorithm produces independent samples.
